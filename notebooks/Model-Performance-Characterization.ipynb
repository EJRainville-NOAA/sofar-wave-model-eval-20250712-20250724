# Model Performance Characterization - 2025-07-12 to 2025-07-24

This work shows plots and calculations for comparing the model performance between four different models (base case, retrov17, gwdupdate, and a bugfix model) run between the dates of 2025-07-12 and 2025-07-24. 

```python
import xarray as xr
import matplotlib.pyplot as plt

%load_ext autoreload
%autoreload 2
from ww3tools import sofar_tools
```

```python
# Load the sofar-model matchup files
data_dir = '/scratch3/NCEPDEV/marine/Edwin.Rainville/sofar-wave-model-eval-20250712-20250724/data/zarr_matchups/'
base_ds = xr.open_dataset(data_dir+'base_case_sofar_matchup_2025-07-12_to_2025-07-24.zarr')
base_ds
```

```python
# Plot a forecast leadtime difference between model and obs hs
fig, ax = plt.subplots()
ax.scatter(base_ds['fcst_hr'], base_ds['model_hs']-base_ds['obs_hs'])
ax.set_xlabel('Forecast Lead Time [hr]')
ax.set_ylabel('Hs Bias (model-obs) [m]')
```

```python
# Plot a forecast leadtime difference between model and obs mean wave period
fig, ax = plt.subplots()
ax.scatter(base_ds['fcst_hr'], base_ds['model_mwp']-base_ds['obs_mwp'])
ax.set_xlabel('Forecast Lead Time [hr]')
ax.set_ylabel('Mean Wave Period Bias (model-obs) [s]')
```

```python
# Plot a forecast leadtime difference between model and obs wind speed
fig, ax = plt.subplots()
ax.scatter(base_ds['fcst_hr'], base_ds['model_wnd']-base_ds['obs_wnd'])
ax.set_xlabel('Forecast Lead Time [hr]')
ax.set_ylabel('Wind Speed Bias (model-obs) [m/s]')
```
